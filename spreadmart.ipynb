```python
# SpendSmart - Expense Analyzer
# Feature 1: Store and display expenses

import numpy as np

expenses = np.array([120, 50, 200, 80, 150,
                     300, 60, 100, 250, 90])

print("===== SpendSmart =====")
print("Your daily expenses:")
print(expenses)
print("Number of transactions:", expenses.size)

# Feature 2: Analyze expenses

total = np.sum(expenses)
average = np.mean(expenses)
highest = np.max(expenses)
lowest = np.min(expenses)

print("===== Expense Statistics =====")
print("Total spending:", total)
print("Average spending:", average)
print("Highest expense:", highest)
print("Lowest expense:", lowest)
# Feature 3: Budget Checker

budget = 1500

print("Your budget:", budget)
print("Your total spending:", total)

if total > budget:
    print("Warning! You exceeded your budget.")
elif total == budget:
    print("You used your entire budget.")
else:
    remaining = budget - total
    print("You are within your budget.")
    print("Money remaining:", remaining)

# Feature 4: Expense Filtering

limit = 100

expensive_transactions = expenses[expenses > limit]

print("===== Expense Filter =====")
print("Filtering expenses above:", limit)
print("Matching expenses:", expensive_transactions)
print("Number of matching transactions:",
      expensive_transactions.size)

# Feature 5: Category-wise Expense Analysis

categories = np.array([
    "Food",
    "Travel",
    "Food",
    "Shopping",
    "Food",
    "Travel",
    "Entertainment",
    "Food",
    "Shopping",
    "Travel"
])

print("===== Category-wise Expenses =====")

unique_categories = np.unique(categories)

for category in unique_categories:
    category_expenses = expenses[categories == category]
    category_total = np.sum(category_expenses)

    print(category, ":", category_total)

# Feature 6: Transaction Summary

print("===== Transaction Summary =====")

for i in range(expenses.size):
    print(
        "Transaction:", i + 1,
        "| Category:", categories[i],
        "| Amount: ₹", expenses[i]
    )
# Feature 7: Spending Percentage

print("===== Spending Percentage =====")

for category in np.unique(categories):
    category_total = np.sum(expenses[categories == category])
    percentage = (category_total / total) * 100

    print(category, ":", round(percentage, 2), "%")

# Feature 8: Largest Transaction Details

max_index = np.argmax(expenses)

print("===== Largest Transaction =====")
print("Amount: ₹", expenses[max_index])
print("Category:", categories[max_index])
print("Transaction number:", max_index + 1)

# Feature 9: Smallest Transaction Details

min_index = np.argmin(expenses)

print("===== Smallest Transaction =====")
print("Amount: ₹", expenses[min_index])
print("Category:", categories[min_index])
print("Transaction number:", min_index + 1)

# Feature 10: Final Expense Report

print("\n========== SPENDSMART REPORT ==========")
print("Total transactions:", expenses.size)
print("Total spending: ₹", total)
print("Average expense: ₹", round(average, 2))
print("Highest expense: ₹", highest)
print("Lowest expense: ₹", lowest)
print("Budget: ₹", budget)
print("Budget remaining: ₹", budget - total)

if total > budget:
    print("Status: Budget exceeded!")
else:
    print("Status: Within budget")

print("=======================================")
```
