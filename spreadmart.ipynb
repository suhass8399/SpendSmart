```python

import numpy as np

print("===== SpendSmart Expense Analyzer =====")

# Get the number of transactions
while True:
    try:
        n = int(input("How many expenses do you want to enter? "))
        if n > 0:
            break
        print("Enter at least 1 transaction.")
    except ValueError:
        print("Please enter a valid whole number.")

# Store transaction details
amounts = []
categories = []
descriptions = []

# Get expense details from the user
for i in range(n):
    print(f"\nExpense {i + 1}")

    # Validate expense amount
    while True:
        try:
            amount = float(input("Enter amount (Rs): "))
            if amount > 0:
                break
            print("Amount must be greater than zero.")
        except ValueError:
            print("Please enter a valid number.")

    category = input("Enter category: ").strip().title()
    description = input("Enter description: ").strip()

    if not category:
        category = "Other"

    if not description:
        description = "No description"

    amounts.append(amount)
    categories.append(category)
    descriptions.append(description)

# Convert numerical data into a NumPy array
expenses = np.array(amounts)

# Get budget
while True:
    try:
        budget = float(input("\nEnter your total budget (Rs): "))
        if budget >= 0:
            break
        print("Budget cannot be negative.")
    except ValueError:
        print("Please enter a valid number.")

# Calculate statistics
total = np.sum(expenses)
average = np.mean(expenses)
highest = np.max(expenses)
lowest = np.min(expenses)

print("\n===== YOUR EXPENSE REPORT =====")

print(f"Total spending: Rs {total:.2f}")
print(f"Average expense: Rs {average:.2f}")
print(f"Highest expense: Rs {highest:.2f}")
print(f"Lowest expense: Rs {lowest:.2f}")

remaining = budget - total

if remaining < 0:
    print(f"Budget exceeded by: Rs {abs(remaining):.2f}")
else:
    print(f"Budget remaining: Rs {remaining:.2f}")

# Display every transaction
print("\n===== TRANSACTION DETAILS =====")

for i in range(n):
    print(
        f"{i + 1}. {descriptions[i]} | "
        f"{categories[i]} | Rs {expenses[i]:.2f}"
    )

# Category-wise analysis
print("\n===== CATEGORY-WISE SPENDING =====")

unique_categories = np.unique(categories)

for category in unique_categories:
    mask = np.array(categories) == category
    category_total = np.sum(expenses[mask])
    percentage = (category_total / total) * 100

    print(
        f"{category}: Rs {category_total:.2f} "
        f"({percentage:.1f}%)"
    )

# Identify largest transaction
largest_index = np.argmax(expenses)
smallest_index = np.argmin(expenses)

print("\n===== TRANSACTION INSIGHTS =====")
print(
    f"Largest: {descriptions[largest_index]} "
    f"(Rs {expenses[largest_index]:.2f})"
)
print(
    f"Smallest: {descriptions[smallest_index]} "
    f"(Rs {expenses[smallest_index]:.2f})"
)

print("\nThank you for using SpendSmart!")
```
